## Question 4

Write a NumPy program to find the most frequent value in an array.

```python
def most_frequent(arr):
    frequent_nums, counts = np.unique(arr, return_counts=True)
    index_of_highest_count = np.argmax(counts)
    return frequent_nums[index_of_highest_count]



def main():
  my_array = np.array([1, 2, 3, 7, 6, 8, 4, 2, 5, 2, 9, 9, 7, 9, 1, 7, 7])
  most_frequent(my_array)
  print(f"the most frequent value in the array is: {most_frequent(my_array)}")

if __name__ == "__main__":
  main()
```

## Question 5

Replace all even numbers in the 2D array with 0.

```python
def replace_even_with_zero(arr):
  arr[arr % 2 == 0] = 0
  return arr

def main():
  arr_2D = np.random.randint(1, 21, size=(3,3))
  print(f"Original Array:\n {arr_2D}")
  print("------------------")
  print(f"Array with Zeros:\n {replace_even_with_zero(arr_2D)}")

if __name__ == "__main__":
  main()
```

## Question 9

Write a NumPy program to create a 5x5 array with random values and find the second-largest value in each column.

```python
arr = np.random.rand(5,5)
arr
```

```python
largest_value_in_col = arr.max(axis=0)
largest_value_in_col
```

```python
index_of_max_value = largest_value_in_col.argmax()
index_of_max_value

```

```python
largest_value_in_col = np.delete(largest_value_in_col, index_of_max_value)
```

```python
second_largest_value = largest_value_in_col.max()
print(f"the second larges value in the array is: {second_largest_value}")
```
