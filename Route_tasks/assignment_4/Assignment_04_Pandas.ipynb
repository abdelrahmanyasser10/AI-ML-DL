## Question 4: Data Visualization
**Problem Statement:**
1. Create a histogram to visualize the distribution of a numerical column.
2. Generate a bar chart to represent the frequency of categories in a categorical column.

```python
plt.hist(df['Avg_Price'], bins=50, color='skyblue', edgecolor='blue')
plt.show()
```

```python
catagory = df['Product_Category'].value_counts(ascending=True)
catagory
```

```python
plt.figure(figsize=(12,6))
plt.barh(catagory.index, catagory.values)
plt.xlabel("Count")
plt.ylabel("Product Category")
plt.title("Distribution of Product Categories")
plt.show()
# len(catagory)

# catagory.shape
# catagory.index
```

```python
plt.figure(figsize=(12,6))
plt.bar(catagory.index, catagory.values)

plt.xticks(rotation=90)
plt.xlabel("Product Category")
plt.ylabel("Count")
plt.title("Distribution of Product Categories")
plt.show()
```
