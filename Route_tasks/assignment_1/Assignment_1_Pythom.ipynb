## Problem 1: Maximum and Minimum of Three Numbers

Write a program to find a maximum and a minimum of three numbers or if the numbers are equal to each other, find if the given number is positive or negative.

**Examples:**
- Input: 6, 2, 8 → Output: minimum is 2 and maximum 8
- Input: 1, 1, 1 → Output: the num is 1

````python
def find_max_min(num1, num2, num3):

  max = num1
  min = num1
  if num1 == num2 == num3:
    print(f"the num is {num1}")
    return

  if num2 > max:
    max = num2
  elif num2 < min:
    min = num2

  if num3 > max:
    max = num3
  elif num3 < min:
    min = num3

  print(f"minimum is {min} and maximum {max}")
  return


user_input = input("Enter three numbers separated by comma: ")

numbers = user_input.split(',')

num1 = int(numbers[0])
num2 = int(numbers[1])
num3 = int(numbers[2])


def main():
  find_max_min(num1, num2, num3)

if __name__ == "__main__":
  main()



````

## Problem 6: String Case Conversion

Given a string, write a python program that convert the string based on the following conditions:
- If lower case letter less than upper case letter convert string to upper case
  - Example: AAAa → AAAA
- If upper case letter less than lower case letter convert string to lower case
  - Example: aaaA → aaaa

````python
def case_conversion(string):
  upper_count = 0
  lower_count = 0
  for char in string:
    if char.isupper():
      upper_count += 1
    elif char.islower():
      lower_count += 1

  if upper_count < lower_count:
    print(string.lower())
  elif lower_count < upper_count:
    print(string.upper())



user_string = input("enter a string: ")

def main():
  case_conversion(user_string)

if __name__ == "__main__":
  main()
````

## Problem 7: Palindrome and Armstrong Number Check

Write a program to accept number from the user and check if the given number is palindrome or Armstrong or not.

**Examples:**
- Palindrome: 12321 (reads the same forwards and backwards)
- Armstrong: 153 (1³ + 5³ + 3³ = 1 + 125 + 27 = 153)

````python
def palindrome_and_armstrong_number_check(number):
  mirror = 0
  armstrong_list = []
  number_copy = number

  for i in range(len(str(number))):
    mirror = (mirror * 10) + (number_copy % 10)
    armstrong_list.append(number_copy % 10)
    number_copy = number_copy // 10

  if mirror == int(number):
    print(f"{int(number)} is a palindrome")

  if sum(i ** len(armstrong_list) for i in armstrong_list) == int(number):
    print(f"{int(number)} is an armstrong number")


user_number = int(input("enter a number: "))

def main():
  palindrome_and_armstrong_number_check(user_number)

if __name__ == "__main__":
  main()
````

## Problem 8: Number to Words Conversion (Bonus)

Write a program to accept number from the user and convert it to words.

**Example:**
- Input: 124 → Output: One hundred and twenty-four

````python
def number_to_words_conversion(number):
    if 0 <= number <= 19:
        return ['zero', 'one', 'two', 'three', 'four', 'five', 'six', 'seven', 'eight', 'nine', 'ten', 'eleven', 'twelve', 'thirteen', 'fourteen', 'fifteen', 'sixteen', 'seventeen', 'eighteen', 'nineteen'][number]
    elif 20 <= number <= 99:
        tens = ['', '', 'twenty', 'thirty', 'forty', 'fifty', 'sixty', 'seventy', 'eighty', 'ninety']
        return tens[number // 10] + ('-' + number_to_words_conversion(number % 10) if number % 10 != 0 else '')
    elif 100 <= number <= 999:
        return number_to_words_conversion(number // 100) + ' hundred' + (' and ' + number_to_words_conversion(number % 100) if number % 100 != 0 else '')
    else:
        return 'Number out of range'

user_number = int(input("enter a number (0-999): "))

while user_number < 0 or user_number > 999:
  user_number = int(input("enter a number between 0 and 999: "))

def main():
  print(number_to_words_conversion(user_number))

if __name__ == "__main__":
  main()
````

Write a program to find the sum of the following series: 1 + 4 – 9 +16 – 25 + 36 ...........n terms

Note: The input is n (number of terms)

**Example:**
- Input: 5 → Output: 1 + 4 – 9 +16 – 25 = -13

````python
def series_sum_calculation(n):
  sum = 1
  for i in range(2, n+1):
    if i % 2 == 0:
      sum += i ** 2
    else:
      sum -= i ** 2
  print(sum)


user_terms = int(input("enter number of terms: "))

def main():
  series_sum_calculation(user_terms)

if __name__ == "__main__":
  main()
````

**Original Tuple:**
```
((10, 10, 10, 12), (30, 45, 56, 45), (81, 80, 39, 32), (1, 2, 3, 4))
```

**Expected Output:**
```
Average value of the numbers of the said tuple of tuples: [30.5, 34.25, 27.0, 23.25]
```

````python
def average_of_tuple_of_tuples(tuple_of_tuples):
  average_of_tuple_of_tuples_list = []
  for i in range(len(tuple_of_tuples)):
    average_of_tuple_of_tuples_list.append(sum(tuple_of_tuples[i]) / len(tuple_of_tuples[i]))
  print(f"Average value of the numbers of the said tuple of tuples: {average_of_tuple_of_tuples_list}")



def main():
  average_of_tuple_of_tuples(((10, 10, 10, 12), (30, 45, 56, 45), (81, 80, 39, 32), (1, 2, 3, 4)))


if __name__ == "__main__":
  main()

````

## Problem 15: Extract Elements with High Frequency

Write a Python program to extract elements with Frequency greater than element K.

**Example:**
- Input: test_list = [4, 6, 4, 3, 3, 4, 3, 4, 6, 6], K = 2
- Output: [4, 3, 6]

````python
def extract_elements_with_high_frequency(test_list, k):
  frequency_list = []
  for i in range(len(test_list)):
    frequency_list.append(test_list.count(test_list[i]))

  result_list = []
  for i in range(len(frequency_list)):
    if frequency_list[i] > k:
      if test_list[i] not in result_list:
        result_list.append(test_list[i])
  print(result_list)


def main():
  extract_elements_with_high_frequency([4, 6, 4, 3, 3, 4, 3, 4, 6, 6], 2)

if __name__ == "__main__":
  main()
````

## Problem 17: Student Dictionary Management

Write a Python program to create a dictionary that stores the names of 3 students, each student has another dictionary with their 'age' and 'grades'. Allow editing of a student's age or grades. The user should choose which student to edit, and then either update the age or modify the list of grades.

**Example Dictionary Structure:**
```python
students = {
    'Alice': {'age': 20, 'grades': [85, 90, 92]},
    'Bob': {'age': 21, 'grades': [75, 80, 78]},
    'Charlie': {'age': 19, 'grades': [95, 97, 93]}
}
```

````python
def student_dictionary_management(students):
    while True:
        print("\nStudent Management System:")
        print("1. View Students")
        print("2. Edit Student Information")
        print("3. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            print("\nStudent List:")
            for student, info in students.items():
                print(f"- {student}: Age - {info['age']}, Grades - {info['grades']}")

        elif choice == '2':
            student_name = input("Enter the name of the student to edit: ")
            if student_name in students:
                print(f"\nEditing {student_name}:")
                print("1. Edit Age")
                print("2. Edit Grades")
                edit_choice = input("Enter your choice: ")

                if edit_choice == '1':
                    try:
                        new_age = int(input(f"Enter new age for {student_name}: "))
                        students[student_name]['age'] = new_age
                        print(f"{student_name}'s age updated.")
                    except ValueError:
                        print("Invalid input. Please enter a number for age.")

                elif edit_choice == '2':
                    grades_input = input(f"Enter new grades for {student_name} (comma-separated numbers): ")
                    try:
                        new_grades = [int(grade.strip()) for grade in grades_input.split(',')]
                        students[student_name]['grades'] = new_grades
                        print(f"{student_name}'s grades updated.")
                    except ValueError:
                        print("Invalid input. Please enter comma-separated numbers for grades.")
                else:
                    print("Invalid edit choice.")
            else:
                print("Student not found.")

        elif choice == '3':
            print("Exiting Student Management System.")
            break

        else:
            print("Invalid choice. Please try again.")


def main():
  students = {
      'Alice': {'age': 20, 'grades': [85, 90, 92]},
      'Bob': {'age': 21, 'grades': [75, 80, 78]},
      'Charlie': {'age': 19, 'grades': [95, 97, 93]}
  }
  student_dictionary_management(students)

if __name__ == "__main__":
  main()
````
